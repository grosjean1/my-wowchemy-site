# Practical session — Gradient descent and quadratic penalization

**Continuous optimization — ENT 305**

**Goal:** solve a constrained optimization problem by minimizing a sequence of *unconstrained* quadratic-penalty functions. This TP replaces the projected-gradient and augmented-Lagrangian parts of the original notebook.

We study the projection of $(0,1)$ onto the line $x=y$:

$$\min_{(x,y)\in\mathbb R^2} f(x,y)=\frac12\bigl[x^2+(y-1)^2\bigr],\qquad h(x,y)=y-x=0.$$

**Questions before coding.** Prove that the unique constrained minimizer is $(\bar x,\bar y)=(1/2,1/2)$. Why is $f$ strictly convex?


```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=6, suppress=True)
```

## 1. Build the quadratic penalty

For $c>0$, define

$$Q_c(x,y)=f(x,y)+\frac c2\,h(x,y)^2.$$

**Exercise 1.** Compute $\nabla f$, $\nabla h$, $\nabla Q_c$ and $\nabla^2 Q_c$ by hand. Complete the functions below, then check the gradient against finite differences. There is **no projection step**: intermediate iterates need not satisfy $x=y$.


```python
def f(x, y):
    return 0.5 * (x*x + (y - 1.0)**2)

def grad_f(x, y):
    return np.array([x, y - 1.0], dtype=float)

def h(x, y):
    return y - x

def grad_h(x, y):
    return np.array([-1.0, 1.0])

def Q(xy, c):
    x, y = np.asarray(xy, dtype=float)
    return f(x, y) + 0.5*c*h(x, y)**2

def grad_Q(xy, c):
    x, y = np.asarray(xy, dtype=float)
    return grad_f(x, y) + c*h(x, y)*grad_h(x, y)

def hessian_Q(c):
    return np.array([[1+c, -c], [-c, 1+c]], dtype=float)

# Numerical gradient check
xy = np.array([0.7, -0.3]); c = 3.0; step = 1e-6
fd = np.array([(Q(xy + step*np.eye(2)[j], c) - Q(xy - step*np.eye(2)[j], c))/(2*step) for j in range(2)])
print('Analytical gradient:', grad_Q(xy, c))
print('Finite differences: ', fd)
assert np.allclose(grad_Q(xy, c), fd, atol=1e-7)
```

## 2. Gradient descent with an Armijo line search

At each iteration, use the descent direction $d_k=-\nabla Q_c(x_k)$ and choose a step $\alpha_k>0$ satisfying

$$Q_c(x_k+\alpha_k d_k)\le Q_c(x_k)+\sigma\alpha_k\nabla Q_c(x_k)^\top d_k,\qquad 0<\sigma<1.$$

**Exercise 2.** Identify the stopping criterion and the line-search condition in the implementation. Compare the trajectories for $c=1$ and $c=20$. Why might a larger penalty make gradient descent slower?


```python
def armijo_step(x, direction, c, alpha0=1.0, shrink=0.5, sigma=1e-4):
    alpha = alpha0
    slope = np.dot(grad_Q(x, c), direction)
    if slope >= 0:
        raise ValueError('The direction is not a descent direction.')
    for _ in range(60):
        if Q(x + alpha*direction, c) <= Q(x, c) + sigma*alpha*slope:
            return alpha
        alpha *= shrink
    raise RuntimeError('Armijo line search did not find a step.')

def gradient_descent(x0, c, tol=1e-8, max_iter=30000):
    x = np.asarray(x0, dtype=float).copy()
    trajectory = [x.copy()]
    for k in range(max_iter):
        grad = grad_Q(x, c)
        if np.linalg.norm(grad) <= tol:
            return np.asarray(trajectory), True
        direction = -grad
        alpha = armijo_step(x, direction, c)
        x = x + alpha*direction
        trajectory.append(x.copy())
    return np.asarray(trajectory), False

for c in [1.0, 20.0]:
    trajectory, converged = gradient_descent([1.5, -0.5], c)
    print(f'c={c:g}: iterations={len(trajectory)-1}, converged={converged}, '
          f'point={trajectory[-1]}, |h|={abs(h(*trajectory[-1])):.3g}')
```

## 3. Analytical solution of the penalized problem

**Exercise 3.** Show that $\nabla Q_c(x_c,y_c)=0$ gives

$$x_c-c(y_c-x_c)=0,\qquad y_c-1+c(y_c-x_c)=0.$$

Add the equations to obtain $x_c+y_c=1$, then prove

$$\boxed{(x_c,y_c)=\left(\frac{c}{1+2c},\frac{1+c}{1+2c}\right).}$$

Why is this the *unique global minimizer* for every $c>0$? Verify the numerical result using gradient descent.


```python
def exact_penalty_solution(c):
    return np.array([c/(1+2*c), (1+c)/(1+2*c)])

for c in [0.5, 1, 5, 20, 100]:
    path, ok = gradient_descent([1.5, -0.5], c, tol=1e-7)
    x_exact = exact_penalty_solution(c)
    print(f'c={c:5g} | GD={path[-1]} | exact={x_exact} | error={np.linalg.norm(path[-1]-x_exact):.2e}')
```

## 4. Effect of the penalty parameter

The solution of the original constrained problem is $\bar z=(1/2,1/2)$. Show that

$$|h(x_c,y_c)|=\frac1{1+2c},\qquad \| (x_c,y_c)-\bar z\|=\frac1{\sqrt2(1+2c)}.$$

**Exercise 4.** Plot these two errors as functions of $c$. What happens as $c\to\infty$? Does any finite $c$ enforce the constraint exactly?


```python
c_values = np.logspace(-1, 3, 100)
feasibility = np.array([abs(h(*exact_penalty_solution(c))) for c in c_values])
distance = np.array([np.linalg.norm(exact_penalty_solution(c)-[0.5,0.5]) for c in c_values])
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.loglog(c_values, feasibility, label=r'$|y_c-x_c|$')
ax.loglog(c_values, distance, label=r'$\|(x_c,y_c)-(1/2,1/2)\|$')
ax.set(xlabel='Penalty parameter c', ylabel='Error', title='Accuracy versus penalty parameter')
ax.grid(True, which='both', alpha=0.3)
ax.legend(); plt.show()
```

## 5. Visualize the iterates

**Exercise 5.** Superimpose the trajectories of gradient descent on level curves of $Q_c$, and draw the line $x=y$. Notice that the algorithm does not project its iterates onto the feasible set.


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
xx, yy = np.meshgrid(np.linspace(-0.6, 1.7, 150), np.linspace(-0.8, 1.7, 150))
for ax, c in zip(axes, [1.0, 20.0]):
    values = 0.5*(xx**2+(yy-1)**2)+0.5*c*(yy-xx)**2
    ax.contour(xx, yy, values, levels=20)
    path, _ = gradient_descent([1.5, -0.5], c)
    ax.plot(path[:, 0], path[:, 1], '.-', markersize=3, label='GD iterates')
    ax.plot([-0.8,1.7], [-0.8,1.7], '--', label='constraint x=y')
    ax.plot(0.5, 0.5, 'o', label='constrained minimizer')
    ax.set(xlim=(-0.6, 1.7), ylim=(-0.8, 1.7), xlabel='x', ylabel='y', title=f'c={c:g}')
    ax.set_aspect('equal', adjustable='box'); ax.legend(fontsize=7)
plt.tight_layout(); plt.show()
```

## 6. Conditioning: the limitation of the method

The Hessian of $Q_c$ is

$$\nabla^2 Q_c=\begin{pmatrix}1+c&-c\\-c&1+c\end{pmatrix}.$$

**Exercise 6.** Show that its eigenvalues are $1$ and $1+2c$. Deduce that $Q_c$ is strongly convex, but that

$$\kappa_2(\nabla^2 Q_c)=1+2c\longrightarrow+\infty.$$

Measure the number of gradient-descent iterations needed as $c$ increases. Explain the compromise between constraint accuracy and numerical conditioning.


```python
test_c = np.array([0.5, 1, 2, 5, 10, 20, 50, 100], dtype=float)
rows = []
for c in test_c:
    path, ok = gradient_descent([1.5, -0.5], c, tol=1e-6, max_iter=20000)
    rows.append((c, np.linalg.cond(hessian_Q(c)), len(path)-1, abs(h(*path[-1])), ok))
print('     c   condition number    GD iterations      |h|      converged')
for c, cond, iters, violation, ok in rows:
    print(f'{c:7.1f} {cond:17.1f} {iters:16d} {violation:12.5g}   {ok}')
```

## 7. Quadratic-penalty algorithm (outer loop)

For a prescribed feasibility tolerance $\varepsilon_{\rm feas}$, increase $c$ progressively, using the previous solution as the initial point for the next unconstrained minimization.

**Exercise 7.** Inspect the effect of changing the growth factor and feasibility tolerance. Why is it usually preferable to increase $c$ gradually rather than choosing an extremely large value at the start?


```python
def penalty_method(x0, c0=0.5, growth=5.0, feasibility_tol=1e-3, max_outer=12):
    x = np.asarray(x0, dtype=float)
    c = c0
    history = []
    for outer in range(max_outer):
        path, converged = gradient_descent(x, c, tol=1e-7, max_iter=30000)
        x = path[-1]
        violation = abs(h(*x))
        history.append((c, x.copy(), violation, len(path)-1, converged))
        if violation <= feasibility_tol:
            break
        c *= growth
    return x, history

x_result, history = penalty_method([1.5, -0.5])
for c, x, violation, iterations, converged in history:
    print(f'c={c:8.2f} x={x} |h|={violation:.3e} GD iterations={iterations} converged={converged}')
print('Final constrained-solution error:', np.linalg.norm(x_result-[0.5,0.5]))
```

## 8. Generalization: affine equality constraints

Consider

$$\min_{x\in\mathbb R^n}\frac12x^\top Hx-q^\top x,\quad Ax=b,\qquad H\succ0.$$

Its quadratic-penalty function is

$$Q_c(x)=\frac12x^\top Hx-q^\top x+\frac c2\|Ax-b\|^2.$$

**Exercise 8.** Differentiate the penalty term and prove that its unique minimizer satisfies

$$\boxed{(H+cA^\top A)x_c=q+cA^\top b.}$$

Show that $H+cA^\top A$ is symmetric positive definite for every $c\ge0$. Explain why linear conjugate gradient can solve this system, whereas the direct KKT matrix is generally indefinite. Does this imply that penalty methods are always numerically preferable?


```python
# Check the affine formula on the same projection example:
# choose x=(x,y), H=I, q=(0,1), A=(-1,1), b=0.
H = np.eye(2)
q = np.array([0.0, 1.0])
A = np.array([[-1.0, 1.0]])
b = np.array([0.0])
for c in [1, 10, 100]:
    matrix = H + c * (A.T @ A)
    rhs = q + c * (A.T @ b)
    x = np.linalg.solve(matrix, rhs)
    print(f'c={c}: solution={x}, eigenvalues={np.linalg.eigvalsh(matrix)}')
```

### Key takeaways

- **Quadratic penalty:** turn an equality-constrained problem into a sequence of unconstrained minimizations $\min Q_c$.
- **No projection:** the intermediate points are generally infeasible.
- **Convergence in this example:** $(x_c,y_c)\to(1/2,1/2)$, but finite $c$ does not enforce $x=y$ exactly.
- **Trade-off:** a larger $c$ improves feasibility but worsens the conditioning of the penalized Hessian.
- **Beyond this TP:** the augmented Lagrangian adds a multiplier estimate to improve feasibility without necessarily taking $c\to\infty$.
